**5. Check Missing Values**

```python
missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values
})

missing_table = missing_table[missing_table["Missing Values"] > 0]

missing_table
```

**6. Check Duplicate Records**

```python
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

**7. Data Types**

```python
data_types = pd.DataFrame({
    "Column": df.columns,
    "Data Type": df.dtypes.astype(str).values
})

data_types
```

**8. Handle Missing Values**

```python
clean_df = df.copy()

numeric_columns = clean_df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(
        clean_df[column].median()
    )


print("Missing values after cleaning:")

print(clean_df.isnull().sum().sum())
```

**9. Categorical Columns**

```python
categorical_columns = clean_df.select_dtypes(
    include=["object", "category"]
).columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

**10. Unique Values**

```python
for column in categorical_columns:
    print("\n", "=" * 50)
    print(column)
    print("=" * 50)
    print(clean_df[column].unique())
```

**11. Season Distribution**

```python
season_counts = clean_df["Season"].value_counts()

print(season_counts)
```

```python
plt.figure(figsize=(8,5))

season_counts.plot(kind="bar")

plt.title("Number of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

**12. Seasonal Average Yield**

```python
season_order = ["Kharif", "Rabi", "Zaid"]

season_yield = (
    clean_df
    .groupby("Season")["Yield_Tonnes_Ha"]
    .mean()
    .reindex(season_order)
)

season_yield
```

```python
plt.figure(figsize=(8,5))

season_yield.plot(kind="bar")

plt.title("Average Agricultural Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

**13. Seasonal Production**

```python
season_production = (
    clean_df
    .groupby("Season")["Production_Tonnes"]
    .mean()
    .reindex(season_order)
)

season_production
```

```python
plt.figure(figsize=(8,5))

season_production.plot(kind="bar")

plt.title("Average Production by Season")
plt.xlabel("Season")
plt.ylabel("Production (Tonnes)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

**14. Seasonal Revenue**

```python
season_revenue = (
    clean_df
    .groupby("Season")["Revenue_INR"]
    .mean()
    .reindex(season_order)
)

season_revenue
```

```python
plt.figure(figsize=(8,5))

(season_revenue / 1000).plot(kind="bar")

plt.title("Average Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue (₹ Thousands)")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

**15. Seasonal Cost**

```python
season_cost = (
    clean_df
    .groupby("Season")["Total_Cost_INR"]
    .mean()
    .reindex(season_order)
)

season_cost
```

**16. Seasonal Profit**

```python
season_profit = (
    clean_df
    .groupby("Season")["Profit_INR"]
    .mean()
    .reindex(season_order)
)

season_profit
```

```python
plt.figure(figsize=(8,5))

(season_profit / 1000).plot(kind="bar")

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (₹ Thousands)")
plt.xticks(rotation=0)

plt.axhline(0, linewidth=1)

plt.tight_layout()
plt.show()
```

**16. Irrigation Method vs Agricultural Performance**

```python
irrigation_summary = (
    clean_df
    .groupby("Irrigation_Method")
    [[
        "Yield_Tonnes_Ha",
        "Profit_INR",
        "Water_Used_m3",
        "Water_Efficiency_t_per_1000m3"
    ]]
    .mean()
    .sort_values(
        "Profit_INR",
        ascending=False
    )
)

# Display the result
irrigation_summary.round(2)
```

```python
plt.figure(figsize=(9,5))

irrigation_summary["Profit_INR"].plot(
    kind="bar"
)

plt.title("Average Profit by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Average Profit (₹)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

**17. State-wise Agricultural Performance**

```python
state_summary = (
    clean_df
    .groupby("State")
    [[
        "Yield_Tonnes_Ha",
        "Production_Tonnes",
        "Revenue_INR",
        "Total_Cost_INR",
        "Profit_INR"
    ]]
    .mean()
    .sort_values(
        "Profit_INR",
        ascending=False
    )
)

state_summary.round(2)
```

```python
plt.figure(figsize=(11,6))

state_summary["Profit_INR"].plot(
    kind="bar"
)

plt.title("Average Profit by State")
plt.xlabel("State")
plt.ylabel("Average Profit (₹)")
plt.xticks(
    rotation=45,
    ha="right"
)

plt.tight_layout()
plt.show()
```

**18. Relationship Between Agricultural Variables**

**a. Select important variables**

```python
correlation_columns = [
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Soil_Moisture_pct",
    "Fertilizer_kg_ha",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]

correlation_matrix = clean_df[
    correlation_columns
].corr()

correlation_matrix.round(2)
```

**b. Create correlation heatmap**

```python
plt.figure(figsize=(12,9))

plt.imshow(
    correlation_matrix,
    aspect="auto"
)

plt.colorbar(
    label="Correlation"
)

plt.xticks(
    range(len(correlation_columns)),
    correlation_columns,
    rotation=70,
    ha="right",
    fontsize=8
)

plt.yticks(
    range(len(correlation_columns)),
    correlation_columns,
    fontsize=8
)

plt.title(
    "Correlation Matrix of Agricultural Variables"
)

plt.tight_layout()
plt.show()
```

**c. Find variables most related to profit**

```python
profit_correlations = (
    correlation_matrix["Profit_INR"]
    .sort_values(
        ascending=False
    )
)

profit_correlations
```

**d. Show positive relationships**

```python
print("Variables positively associated with Profit:")
print("=" * 50)

positive_relationships = profit_correlations[
    profit_correlations > 0
]

display(
    positive_relationships.round(3)
)
```
